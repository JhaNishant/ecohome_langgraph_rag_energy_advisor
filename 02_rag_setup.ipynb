# EcoHome Energy Advisor: Knowledge Base Setup

This notebook builds the local search index that gives the advisor grounded energy saving tips.

```python
from dotenv import load_dotenv
from tools import build_energy_tip_vectorstore, load_energy_tip_documents, search_energy_tips

load_dotenv()
```

```python
documents = load_energy_tip_documents()
sources = sorted({document.metadata['source'] for document in documents})
assert len(sources) == 7
print('Indexed source documents:')
for source in sources:
    print('-', source)
```

```python
vectorstore = build_energy_tip_vectorstore(rebuild=True)
print('Chroma knowledge base is ready.')
```

```python
test_queries = [
    'EV charging with solar power',
    'heat pump thermostat settings',
    'dishwasher automation',
    'home battery scheduling',
    'seasonal energy planning',
]
for query in test_queries:
    result = search_energy_tips.invoke({'query': query, 'max_results': 2})
    print(f"\n{query}")
    for tip in result['tips']:
        print(f"  {tip['source']}: {tip['content'][:140]}...")
```

## What this means

The advisor can now use practical source material instead of relying only on general model knowledge. Each search result keeps its source filename so answers can cite the tip it used.
